```
/// <summary>
/// Универсальная функция для вывода массива строк в консоль
/// </summary>
void PrintShape(string[] shape)
{
    Console.WriteLine(string.Join(Environment.NewLine, shape));
}

/// <summary>
/// Разделитель между тестами для наглядности
/// </summary>
void PrintSeparator()
{
    Console.WriteLine(new string('-', 40));
}
```

```
/// <summary>
/// Получить прямоугольник с настройкой заливки и толщины рамки
/// </summary>
string[] GetRectangle(int width, int height, char symbol = '*', char bg = ' ', bool isFill = false, int border = 1)
{
    var result = new List<string>();
    for (int r = 0; r < height; r++)
    {
        var row = new char[width];
        for (int c = 0; c < width; c++)
        {
            bool isBorder = r < border || r >= height - border || c < border || c >= width - border;
            row[c] = (isFill || isBorder) ? symbol : bg;
        }
        result.Add(new string(row));
    }
    return result.ToArray();
}

void PrintRectangle(int width, int height, char symbol = '*', char bg = ' ', bool isFill = false, int border = 1)
{
    PrintShape(GetRectangle(width, height, symbol, bg, isFill, border));
}
```

```
PrintRectangle(10, 5);
PrintSeparator();
PrintRectangle(10, 5, isFill: true, symbol: '#');
PrintSeparator();
PrintRectangle(10, 5, border: 2, symbol: '@', bg: '.');
PrintSeparator();
```

```
/// <summary>
/// orientation: 1 - левый верх, 2 - правый верх, 3 - левый низ, 4 - правый низ
/// </summary>
string[] GetTriangle(int size, char symbol = '*', bool isFill = true, int orientation = 1)
{
    var result = new List<string>();
    for (int r = 0; r < size; r++)
    {
        int len = orientation <= 2 ? size - r : r + 1;
        int pad = size - len;
        
        string line = isFill || len <= 2 
            ? new string(symbol, len) 
            : symbol + new string(' ', len - 2) + symbol;
            
        line = orientation switch
        {
            1 or 3 => line,
            _ => new string(' ', pad) + line
        };
        result.Add(line);
    }
    return result.ToArray();
}

void PrintTriangle(int size, char symbol = '*', bool isFill = true, int orientation = 1)
{
    PrintShape(GetTriangle(size, symbol, isFill, orientation));
}
```

```
PrintTriangle(6, orientation: 1);
PrintSeparator();
PrintTriangle(6, orientation: 2);
PrintSeparator();
PrintTriangle(6, orientation: 3, isFill: false);
PrintSeparator();
PrintTriangle(6, orientation: 4, isFill: false);
PrintSeparator();
```

```
string[] GetRhombus(int size, char symbol = '*', bool isFill = true)
{
    var result = new List<string>();
    int height = size * 2 - 1;
    for (int r = 0; r < height; r++)
    {
        int dist = Math.Abs(r - (size - 1));
        int stars = size - dist;
        int spaces = size - 1 - dist;
        
        string body = isFill || stars <= 1 
            ? new string(symbol, stars * 2 - 1) 
            : symbol + new string(' ', (stars - 2) * 2 + 1) + symbol;
            
        result.Add(new string(' ', spaces) + body);
    }
    return result.ToArray();
}

void PrintRhombus(int size, char symbol = '*', bool isFill = true) => PrintShape(GetRhombus(size, symbol, isFill));

string[] GetHourglass(int height, char symbol = '*')
{
    var result = new List<string>();
    int mid = height / 2;
    for (int r = 0; r < height; r++)
    {
        int dist = r <= mid ? r : height - 1 - r;
        int stars = Math.Max(1, height - 2 * dist);
        int spaces = dist;
        result.Add(new string(' ', spaces) + new string(symbol, stars) + new string(' ', spaces));
    }
    return result.ToArray();
}

void PrintHourglass(int height, char symbol = '*') => PrintShape(GetHourglass(height, symbol));
```

```
PrintRhombus(5);
PrintSeparator();
PrintRhombus(5, isFill: false, symbol: '$');
PrintSeparator();
PrintHourglass(7);
PrintSeparator();
PrintHourglass(6, symbol: '~');
PrintSeparator();
```

```
/// <summary>
/// Зебра: чередующиеся полосы
/// </summary>
string[] GetZebra(int width, int height, char symbol = '#', char bg = ' ', int thickness = 2, bool vertical = false)
{
    var result = new List<string>();
    for (int r = 0; r < height; r++)
    {
        var row = new char[width];
        for (int c = 0; c < width; c++)
        {
            int index = vertical ? c / thickness : r / thickness;
            row[c] = (index % 2 == 0) ? symbol : bg;
        }
        result.Add(new string(row));
    }
    return result.ToArray();
}

void PrintZebra(int width, int height, char symbol = '#', char bg = ' ', int thickness = 2, bool vertical = false)
    => PrintShape(GetZebra(width, height, symbol, bg, thickness, vertical));
```

```
PrintZebra(20, 10, thickness: 2);
PrintSeparator();
PrintZebra(20, 10, thickness: 1);
PrintSeparator();
PrintZebra(20, 10, thickness: 2, vertical: true);
PrintSeparator();
```

```
/// <summary>
/// Змейка: извилистый путь с чередованием сторон поворота
/// </summary>
string[] GetSnake(int width, int height, char symbol = '#', char bg = ' ', int thickness = 2, bool vertical = false)
{
    var grid = new char[height, width];
    for(int r = 0; r < height; r++) 
        for(int c = 0; c < width; c++) 
            grid[r,c] = bg;

    if (!vertical)
    {
        int r = 0;
        int turnSide = 0; // 0 - право, 1 - лево
        
        while (r < height)
        {
            for (int dr = 0; dr < thickness && r + dr < height; dr++)
                for (int c = 0; c < width; c++)
                    grid[r + dr, c] = symbol;
            r += thickness;

            if (r < height)
            {
                int col = (turnSide == 0) ? width - thickness : 0;
                for (int dr = 0; dr < thickness && r + dr < height; dr++)
                    for (int dc = 0; dc < thickness && col + dc < width; dc++)
                        grid[r + dr, col + dc] = symbol;
                r += thickness;
                turnSide = 1 - turnSide;
            }
        }
    }
    else
    {
        int c = 0;
        int turnSide = 0; // 0 - низ, 1 - верх
        
        while (c < width)
        {
            for (int dc = 0; dc < thickness && c + dc < width; dc++)
                for (int r = 0; r < height; r++)
                    grid[r, c + dc] = symbol;
            c += thickness;

            if (c < width)
            {
                int row = (turnSide == 0) ? height - thickness : 0;
                for (int dc = 0; dc < thickness && c + dc < width; dc++)
                    for (int dr = 0; dr < thickness && row + dr < height; dr++) 
                        grid[row + dr, c + dc] = symbol;
                c += thickness;
                turnSide = 1 - turnSide;
            }
        }
    }

    var result = new List<string>();
    for (int r = 0; r < height; r++)
    {
        var rowChars = new char[width];
        for (int c = 0; c < width; c++) rowChars[c] = grid[r, c];
        result.Add(new string(rowChars));
    }
    return result.ToArray();
}

void PrintSnake(int width, int height, char symbol = '#', char bg = ' ', int thickness = 2, bool vertical = false)
    => PrintShape(GetSnake(width, height, symbol, bg, thickness, vertical));
```

```
/// <summary>
/// Вложенные прямоугольники
/// </summary>
string[] GetNestedRectangles(int width, int height, char symbol = '#', char bg = ' ', int thickness = 1)
{
    var result = new List<string>();
    for (int r = 0; r < height; r++)
    {
        var row = new char[width];
        for (int c = 0; c < width; c++)
        {
            int dist = Math.Min(Math.Min(r, c), Math.Min(height - 1 - r, width - 1 - c));
            int layer = dist / thickness;
            
            row[c] = (layer % 2 == 0) ? symbol : bg;
        }
        result.Add(new string(row));
    }
    return result.ToArray();
}

void PrintNestedRectangles(int width, int height, char symbol = '#', char bg = ' ', int thickness = 1)
    => PrintShape(GetNestedRectangles(width, height, symbol, bg, thickness));
```

```
PrintNestedRectangles(20, 10, thickness: 1);
PrintSeparator();
PrintNestedRectangles(20, 10, thickness: 2);
PrintSeparator();
```

```
/// <summary>
/// Улиточная спираль
/// </summary>
string[] GetSnailSpiral(int width, int height, char symbol = '#', char bg = ' ', bool clockwise = true)
{
    if (width <= 0 || height <= 0) return Array.Empty<string>();

    char[][] grid = new char[height][];
    for (int i = 0; i < height; i++)
    {
        grid[i] = new char[width];
        Array.Fill(grid[i], bg);
    }

    int[] dr = clockwise ? [0, 1, 0, -1] : [1, 0, -1, 0];
    int[] dc = clockwise ? [1, 0, -1, 0] : [0, 1, 0, -1];

    int r = 0, c = 0;
    int dir = 0;

    grid[r][c] = symbol;

    bool CanMove(int tr, int tc)
    {
        if (tr < 0 || tr >= height || tc < 0 || tc >= width) return false;
        
        if (grid[tr][tc] == symbol) return false;

        int neighbors = 0;
        if (tr > 0 && grid[tr - 1][tc] == symbol) neighbors++;
        if (tr < height - 1 && grid[tr + 1][tc] == symbol) neighbors++;
        if (tc > 0 && grid[tr][tc - 1] == symbol) neighbors++;
        if (tc < width - 1 && grid[tr][tc + 1] == symbol) neighbors++;
        return neighbors <= 1;
    }

    while (true)
    {
        int nr = r + dr[dir];
        int nc = c + dc[dir];

        if (CanMove(nr, nc))
        {
            r = nr;
            c = nc;
            grid[r][c] = symbol;
        }
        else
        {
            int nextDir = (dir + 1) % 4;
            nr = r + dr[nextDir];
            nc = c + dc[nextDir];

            if (CanMove(nr, nc))
            {
                dir = nextDir;
                r = nr;
                c = nc;
                grid[r][c] = symbol;
            }
            else
            {
                break;
            }
        }
    }

    var result = new string[height];
    for (int i = 0; i < height; i++)
    {
        result[i] = new string(grid[i]);
    }
    return result;
}

void PrintSnailSpiral(int width, int height, char symbol = '#', char bg = ' ', bool clockwise = true)
    => PrintShape(GetSnailSpiral(width, height, symbol, bg, clockwise));
```

```
PrintSeparator();
PrintSnailSpiral(15, 11, '█', ' ', clockwise: true);
PrintSeparator();
PrintSnailSpiral(15, 11, '█', ' ', clockwise: false);
PrintSeparator();
PrintSnailSpiral(15, 15, '#', '.');
```
