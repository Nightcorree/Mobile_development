```
delegate void Work();

void Hello() => Console.WriteLine("AAAAAAA");
void Hellolallal() => Console.WriteLine("BBBBBBBB");


Work work;
work += Hello;
work += Hellolallal;
work();
```

```
int Add(int x, int y) => x + y;
int Sub(int x, int y) => x + y;
int Mul(int x, int y) => x + y;

delegate int MathOperation(int x, int y);

MathOperation calc;
calc = Add;
calc(2, 3)
```

```
int Add(int x, int y) => x + y;
int Sub(int x, int y) => x + y;
int Mul(int x, int y) => x + y;

enum MathOperator
{
    Add,
    Sub,
    Mul
}
delegate int MathOperation(int x, int y);
Dictionary<MathOperator, MathOperation> calc = new()
{
    [MathOperator.Add] = Add,
    [MathOperator.Sub] = Sub,
    [MathOperator.Mul] = Mul
};
calc[MathOperator.Add](2, 3)
```
