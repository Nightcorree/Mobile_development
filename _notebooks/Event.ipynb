```
class Person
{
    public delegate void MyHandler(string newValue);
    public event MyHandler OnNameChanged;
    private string name;
    public string Name
    {
        get => name;
        set => OnNameChanged.Invoke(name = value);
    }
}
var p = new Person();
p.OnNameChanged += st => Console.WriteLine($"Новое значение {st}") ;
p.Name = "Юра";
p.Name = "Паша";
p.Name = "Максим";
```
