# Массив (Array)

```
string[] cities = {"Тула", "Москва", "Санкт-Петербург", "Новосибирск", "Екатеринбург"};
Console.WriteLine(cities[0]); // Выводит "Тула"
Console.WriteLine(cities[^1]); // Выводит "Екатеринбург"
Console.WriteLine(cities.Length); // Выводит "5"
```

Как по значению элемента найти его индекс в массиве?

```
Console.WriteLine(Array.IndexOf(cities, "Москва")); // Выводит "1"
Console.WriteLine(Array.IndexOf(cities, "Казань1")); // Выводит "-1"
```

Диапозоны

```
Range range = 1..3;
Console.WriteLine(string.Join(", ", cities[range])); 
```

Распространение

```
int[] a1 = [1, 2, 3];
int[] a2 = [4, 5, 6];
int[] a3 = [..a1, ..a2 ];

 // Выводит "1, 2, 3, 4, 5, 6"
```
